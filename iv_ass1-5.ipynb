```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
from tkinter import Tk, filedialog

# Open file selection window
root = Tk()
root.withdraw()

image_path = filedialog.askopenfilename(
    title="Select an Image",
    filetypes=[("Image Files", "*.jpg *.jpeg *.png")]
)

print("Selected Image:", image_path)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Read selected image
img = Image.open(image_path).convert("RGB")

# Convert image into NumPy array
img_array = np.array(img)

# Display original image
plt.imshow(img)
plt.title("Original RGB Image")
plt.axis("off")
plt.show()
```

```python
R = img_array[:, :, 0]
G = img_array[:, :, 1]
B = img_array[:, :, 2]

gray_mean = (R + G + B) / 3

gray_mean = gray_mean.astype(np.uint8)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original RGB Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gray_mean, cmap="gray")
plt.title("Grayscale - Mean Method")
plt.axis("off")

plt.show()

Image.fromarray(gray_mean).save("grayscale_mean.jpg")

print("Grayscale image saved successfully!")
```

```python
# Case 2: RGB to Grayscale using User-Defined Weightage

# Take weightage from user
wr = float(input("Enter Red weightage: "))
wg = float(input("Enter Green weightage: "))
wb = float(input("Enter Blue weightage: "))

# Check whether weightage is valid
if (0 <= wr <= 1 and 0 <= wg <= 1 and 0 <= wb <= 1
    and np.isclose(wr + wg + wb, 1.0)):

    # Calculate weighted grayscale
    gray_weighted = (R * wr) + (G * wg) + (B * wb)

    # Convert into uint8 format
    gray_weighted = np.clip(gray_weighted, 0, 255).astype(np.uint8)

    # Display original and grayscale images
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(img)
    plt.title("Original RGB Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(gray_weighted, cmap="gray")
    plt.title("Grayscale - Weighted Method")
    plt.axis("off")

    plt.show()

    # Save grayscale image
    Image.fromarray(gray_weighted).save("grayscale_weighted.jpg")

    print("Weighted grayscale image saved successfully!")

else:
    print("Invalid weightage!")
    print("Each weight must be between 0 and 1.")
    print("Sum of all weightages must be equal to 1.")
```
